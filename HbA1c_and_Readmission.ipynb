# Hospital Readmission Risk Analysis
## Project Overview
This project was developed for a hospital group seeking to reduce preventable readmissions and better allocate post-discharge resources. Using ten years of patient readmission data, the goal was to identify which clinical and procedural variables — such as initial diagnosis, number of procedures performed, HbA1c measurements, and other patient-level factors — are most predictive of a patient returning to the hospital after discharge.

A key focus of the analysis was exploring the relationship between diabetes management and readmission risk. HbA1c, a marker of long-term blood glucose control, was examined to determine whether glycemic status at the time of discharge is associated with higher rates of readmission — providing insight into how uncontrolled or poorly managed diabetes may contribute to adverse outcomes post-discharge.

The analysis produced a readmission probability model that enables care teams to prioritize follow-up outreach toward patients at elevated risk, supporting more proactive, targeted interventions and ultimately improving patient outcomes.
